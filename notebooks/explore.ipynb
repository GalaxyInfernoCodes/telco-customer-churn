# Telco Customer Churn Exploration
Exploring the dataset located at `../artifacts/Telco-Customer-Churn.csv`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
telco_full = pd.read_csv('../artifacts/Telco-Customer-Churn.csv')
telco_full.head()
```

```python
# Display dataframe info
telco_full.info()
```

```python
# Display summary statistics
telco_full.describe()
```

## Plot distribution of features

```python

for column in telco_full.columns:
    plt.figure(figsize=(10, 4))
    sns.histplot(telco_full[column])
    plt.title(f'Distribution of {column}')
    plt.show()
```
