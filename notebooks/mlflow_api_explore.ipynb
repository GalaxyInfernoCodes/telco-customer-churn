```python
import mlflow   
```

```python
mlflow.set_tracking_uri(f"sqlite:///../mlflow.db")
```

```python
experiment_name = "churn_baseline_balanced"
experiment = mlflow.get_experiment_by_name(experiment_name)
experiment
```

```python

if experiment:
    experiment_id = experiment.experiment_id

    runs_df = mlflow.search_runs(
        experiment_ids = [experiment_id],
        max_results=10,
        order_by = ["metrics.val_pr_auc DESC"]
    )

    display(runs_df)
else:
    print("Experiment not found.")
```

```python
from google import genai
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)

response = client.models.generate_content(
    model = "gemini-3.5-flash",
    contents = f"Suggest new hyperparameters for my catboost training, here are the best 10 recent runs by validation pr_auc: {runs_df}"
)

print(response.text)
```
